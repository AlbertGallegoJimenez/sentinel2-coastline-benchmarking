```python
# Comparativa Salinas vs Puntal
# En las siguientes celdas se comparan las métricas de brillo/luminosidad entre sitios,
# su relación con hs/tp/tide y la diferencia relativa (%) entre playas.
```

```python
# Preparar datos y definir un subconjunto de métricas más interpretable
df = results.copy()
df["image_datetime"] = pd.to_datetime(df["image_datetime"], errors="coerce")

site_col = "site"
ocean_cols = ["hs", "tp", "tide"]

exclude_cols = {"source", "site", "image_name", "image_datetime", *ocean_cols, "roi_pixels"}
metric_cols = [c for c in df.columns if c not in exclude_cols]

# Asegurar que las métricas sean numéricas
for col in metric_cols + ocean_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Subconjunto recomendado: brillo medio + contraste espectral (sin métricas de desviación)
focus_metrics = [
    "mean_rgb_image_mean",
    "luminance_image_mean",
    "r_minus_b_mean",
]
focus_metrics = [m for m in focus_metrics if m in metric_cols]

print("Sitios:", sorted(df[site_col].dropna().unique().tolist()))
print("Métricas disponibles:", metric_cols)
print("Métricas recomendadas:", focus_metrics)
display(df[[site_col, "image_datetime", *ocean_cols, *focus_metrics]].head())
```

```python
# Descripción estadística de las variables metoceánicas (hs, tp y tide) por sitio
# Visualizar una figura con tres subplots expresando histogramas+PDFs usando el color para diferenciar sitios. Esto nos ayudará a entender si las condiciones oceánicas son similares o diferentes entre sitios, lo cual podría influir en las métricas de brillo/luminosidad. Además, se pueden incluir estadísticas descriptivas (media, mediana, desviación estándar) para cada variable y sitio.

fig, axes = plt.subplots(1, 3, figsize=(18, 5), constrained_layout=True)
for i, ocean_var in enumerate(ocean_cols):
    ax = axes[i]
    sns.histplot(data=df, x=ocean_var, hue=site_col, ax=ax, kde=True, palette="Set2", alpha=0.6)
    ax.set_title(f"Distribución de {ocean_var}")
    ax.set_xlabel(ocean_var)
    ax.set_ylabel("Frecuencia")
```

```python
import matplotlib.patheffects as pe

focus_metrics = ['mean_rgb_image_mean', 'r_minus_b_mean']
focus_metrics_labels = {
    'mean_rgb_image_mean': 'Mean Visible Reflectance (RGB)',
    'r_minus_b_mean': 'Red-Blue Differential (R-B)'
}

# Comparación directa entre sitios para métricas recomendadas
plot_df = df.melt(id_vars=[site_col], value_vars=focus_metrics,
                  var_name="metric", value_name="value").dropna()

# Reorder entries so "Puntal" appears before "Salinas"
#plot_df[site_col] = pd.Categorical(plot_df[site_col], categories=["puntal", "salinas"], ordered=True)

n_metrics = len(focus_metrics)
ncols = 2
nrows = int(np.ceil(n_metrics / ncols))

fig, axes = plt.subplots(nrows, ncols, figsize=(8, 4), constrained_layout=True)
axes = np.array(axes).reshape(-1)

for i, metric in enumerate(focus_metrics):
    ax = axes[i]
    d = plot_df[plot_df["metric"] == metric]
    sns.boxplot(data=d, x=site_col, y="value", ax=ax, hue=site_col, hue_order=["puntal", "salinas"],
                showfliers=False,
                boxprops=dict(edgecolor="black", alpha=0.7), 
                medianprops=dict(color="white", linewidth=2.5),
                whiskerprops=dict(color="black"),
                capprops=dict(color="black"),
                legend=False)
    sns.stripplot(data=d, x=site_col, y="value", hue_order=["puntal", "salinas"], legend=False,
                  palette=[palette[1], palette[0]],
                  ax=ax, color="0.2", size=6, jitter=True, alpha=0.7)
    ax.set_title(focus_metrics_labels.get(metric, metric), fontweight="bold")
    ax.set_xlabel("Study Area", fontweight="bold")
    ax.set_ylabel("Metric Value", fontweight="bold")
    ax.set_xticklabels([s.get_text().capitalize() for s in ax.get_xticklabels()])
    sns.despine(ax=ax)

# Show median values label on the plot
medians = plot_df.groupby(['metric', site_col])['value'].median().reset_index()
for i, row in medians.iterrows():
    metric = row['metric']
    site = row[site_col]
    median_value = row['value']
    ax = axes[focus_metrics.index(metric)]
    x_pos = 1 if site == "puntal" else 0
    ax.text(x_pos, median_value, f"{median_value:.2f}", ha='center', va='center', fontweight='bold',
            path_effects=[pe.withStroke(linewidth=3, foreground="white")])

for j in range(i + 1, len(axes)):
    axes[j].axis("off")

#fig.suptitle("Distribución por sitio (métricas seleccionadas)", y=1.08)
plt.show()

# Export figure
out_dir = os.path.join("..", "results", "figures", "sand_reflectance_analysis")
os.makedirs(out_dir, exist_ok=True)
fig.savefig(os.path.join(out_dir, "figure_brightness_comparison_by_site.png"), dpi=300, bbox_inches="tight")
```

```python
# Test estadístico entre playas para métricas recomendadas
from scipy.stats import mannwhitneyu

def cliffs_delta(x, y):
    # Tamaño de efecto no paramétrico: probabilidad de superioridad entre grupos
    x = np.asarray(x)
    y = np.asarray(y)
    gt = sum(np.sum(xi > y) for xi in x)
    lt = sum(np.sum(xi < y) for xi in x)
    return (gt - lt) / (len(x) * len(y))

def bh_fdr(p_values):
    p = np.asarray(p_values, dtype=float)
    n = len(p)
    order = np.argsort(p)
    ranked = p[order]
    q = ranked * n / (np.arange(1, n + 1))
    q = np.minimum.accumulate(q[::-1])[::-1]
    q = np.clip(q, 0, 1)
    out = np.empty_like(q)
    out[order] = q
    return out

sites = sorted(df[site_col].dropna().unique().tolist())
if len(sites) != 2:
    print("Se requieren exactamente dos sitios para esta comparación.")
else:
    s1, s2 = sites[0], sites[1]
    rows = []
    for metric in focus_metrics:
        x = df.loc[df[site_col] == s1, metric].dropna().values
        y = df.loc[df[site_col] == s2, metric].dropna().values
        if len(x) < 3 or len(y) < 3:
            continue
        stat, p = mannwhitneyu(x, y, alternative="two-sided")
        delta = cliffs_delta(x, y)
        rows.append({
            "metric": metric,
            f"median_{s1}": np.median(x),
            f"median_{s2}": np.median(y),
            f"median_diff_{s1}_minus_{s2}": np.median(x) - np.median(y),
            "mannwhitney_u": stat,
            "p_value": p,
            "cliffs_delta": delta,
            "abs_cliffs_delta": abs(delta),
        })

    stats_df = pd.DataFrame(rows)
    if not stats_df.empty:
        stats_df["p_value_fdr_bh"] = bh_fdr(stats_df["p_value"].values)
        stats_df["significant_0_05"] = stats_df["p_value_fdr_bh"] < 0.05
        stats_df = stats_df.sort_values(["significant_0_05", "abs_cliffs_delta", "p_value_fdr_bh"], ascending=[False, False, True])
        display(stats_df.style.format({
            "p_value": "{:.3e}",
            "p_value_fdr_bh": "{:.3e}",
            "cliffs_delta": "{:+.3f}",
            "abs_cliffs_delta": "{:.3f}",
            f"median_{s1}": "{:.3f}",
            f"median_{s2}": "{:.3f}",
            f"median_diff_{s1}_minus_{s2}": "{:+.3f}",
        }))

        plt.figure(figsize=(9, 0.7 * len(stats_df) + 2))
        sns.barplot(data=stats_df, x="cliffs_delta", y="metric", hue="significant_0_05", palette="Set1")
        plt.axvline(0, color="black", linewidth=1)
        plt.xlabel("Cliff's delta (dirección y magnitud de diferencia)")
        plt.ylabel("Métrica")
        plt.title(f"Tamaño de efecto {s1} vs {s2} (Mann-Whitney + FDR)")
        plt.legend(title="Significativa (FDR<0.05)")
        plt.tight_layout()
        plt.show()
    else:
        print("No hay datos suficientes para ejecutar los test.")
```
